• Response variable: CurrentCigaretteUse

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.stats.proportion import proportions_ztest, confint_proportions_2indep

# Load data
df = pd.read_csv('../data/raw/YRBS_2007.csv')
```

```python
# Save the original column order for later restoration
original_cols = df.columns.tolist()  

# Remove missing values
df = df.dropna(subset=["WhatIsYourSex", "CurrentCigaretteUse"])

# Recode gender into 0/1
df["WhatIsYourSex"] = df["WhatIsYourSex"].map({
    1: 0,  # Female
    2: 1   # Male
})

# Recoding
df["CurrentCigaretteUse"] = pd.to_numeric(df["CurrentCigaretteUse"], errors="coerce")

df["CurrentCigaretteUse"] = df["CurrentCigaretteUse"].apply(
    lambda x: 1 if 2 <= x <= 7 else (0 if x == 1 else np.nan)
)

# Remove NaN
df = df.dropna(subset=["WhatIsYourSex", "CurrentCigaretteUse"])

# Type conversion
df["WhatIsYourSex"] = df["WhatIsYourSex"].astype(int)
df["CurrentCigaretteUse"] = df["CurrentCigaretteUse"].astype(int)

# Reset index starting from 1
df.index = np.arange(1, len(df) + 1)

# Display version (layout only)
show_cols = ["WhatIsYourSex", "CurrentCigaretteUse"] + \
            [c for c in df.columns if c not in ["WhatIsYourSex", "CurrentCigaretteUse"]]

df_show = df[show_cols]

# Export CSV
df = df[original_cols]   
df.to_csv("../data/processed/yrbs_cleaned_and_recoded.csv", index=False)

print("yrbs_cleaned_and_recoded.csv has been successfully exported.\nResults are shown below:")

# Display settings
pd.set_option("display.max_columns", 10)
pd.set_option("display.max_rows", None)

# Show results
df_show
```
